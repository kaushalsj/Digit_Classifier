```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.random.rand(100, 784)

y = np.random.randint(0, 10, 100)

print(X.shape)
print(y.shape)
```

```python
W1 = np.random.randn(784, 10)
b1 = np.zeros((1, 10))

W2 = np.random.randn(10, 10)
b2 = np.zeros((1, 10))
```

```python
def relu(x):
    return np.maximum(0, x)
```

```python
def softmax(x):
    exp = np.exp(x)
    return exp / np.sum(exp, axis=1, keepdims=True)
```

```python
Z1 = np.dot(X, W1) + b1
A1 = relu(Z1)

Z2 = np.dot(A1, W2) + b2

predictions = softmax(Z2)
```

```python
predicted_labels = np.argmax(predictions, axis=1)

accuracy = np.mean(predicted_labels == y)

print("Accuracy:", accuracy)
```

```python
loss = -np.mean(np.log(np.max(predictions, axis=1)))

print(loss)
```

```python
losses = [2.5, 2.2, 2.0, 1.8, 1.6]
```

```python
plt.plot(losses)
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

accuracies = [accuracy]

plt.plot(accuracies, marker='o')
plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.show()
losses = [2.5, 2.2, 2.0, 1.8, 1.6]

```
